```python
from robovast.common.analysis import (CampaignDataError, config_file,
                                      get_behavior_info, read_runs, read_table)
# Public, but not re-exported from the package, so it comes from the submodule. Imported
# rather than reimplemented: which clock a pose may be differentiated against is the
# library's rule, and a second copy of it here is a copy that goes stale.
from robovast.common.analysis.metrics import pose_time_base
import pandas as pd
# `robovast_nav.map_visualizer`, not `robovast_nav.gui`: the gui package went with the Qt
# desktop tools, and the matplotlib visualizer moved up beside its siblings.
from robovast_nav.map_visualizer import MapVisualizer
import matplotlib.pyplot as plt
import matplotlib.cm as cm
import numpy as np
import os
DATA_DIR = ''

# These notebooks serve every .vast in this directory, and those split into two families:
# five scenarios drive `robot.follow_waypoints`, and scenario_nav_through_poses.osc drives
# scenario_nav_through_poses.osc drive `robot.nav_through_poses`. The behaviour is
# recorded under `<actor type>.<action>`, so a name hardcoded to either family leaves the
# other with empty duration panels and no error -- read it from the tree instead.
NAV_ACTIONS = ('nav_through_poses', 'follow_waypoints', 'nav_to_pose')


def navigation_behavior(behaviors):
    """The navigation behaviour's name as this campaign recorded it."""
    names = set(behaviors['behavior_name'].dropna())
    for action in NAV_ACTIONS:
        found = sorted(n for n in names if n.rsplit('.', 1)[-1] == action)
        if found:
            return found[0]
    raise SystemExit(
        f"No navigation behaviour here. Looked for {', '.join(NAV_ACTIONS)}; the tree "
        f"recorded: {', '.join(sorted(n for n in names if '.' in n)) or '(none)'}.")


# One read per table, so the message names the one that is missing. Together in a single
# try, a campaign with poses but no behaviour tree -- execution.bt_log off, or an execution
# image whose scenario_execution predates --bt-log -- reported "no postprocessed data",
# which is a different problem with a different fix.
try:
    df = read_table(DATA_DIR, "poses", with_params=True,
                    require=['frame', 'timestamp', 'position.x', 'position.y',
                             'orientation.yaw'])
except CampaignDataError as e:
    print(e)
    raise SystemExit("This configuration has no postprocessed data to analyse.")

try:
    df_behaviors = read_table(DATA_DIR, "behaviors",
                              require=['behavior_name', 'status_name', 'timestamp'])
except CampaignDataError as e:
    # Every panel from here on is keyed off the navigation behaviour, so there is no
    # partial view worth rendering -- but the reason is not the one above.
    print(e)
    raise SystemExit(
        "This configuration recorded no behaviour tree, so its navigation cannot be timed or "
        "named. See the message above for the two ways that happens.")

if df.empty:
    raise SystemExit("No poses recorded for any run of this configuration.")

runs = df['run_id'].unique()
# A run that died before its first transform contributes no poses, and every panel below
# is keyed off the poses -- so without this the notebook analyses 4 of 5 runs and reports
# "4 runs", which reads as the configuration's repetition count.
df_runs = read_runs(DATA_DIR)
missing = sorted(set(df_runs['run_id']) - set(runs))
if missing:
    outcomes = df_runs.set_index('run_id')['status'].to_dict()
    print(f"{len(missing)} of {len(df_runs)} run(s) recorded no poses and are absent from "
          f"everything below: "
          + ", ".join(f"run {r} ({outcomes.get(r, 'unknown')})" for r in missing))
# One colour per run, used by every per-run panel below.
colors = cm.rainbow(np.linspace(0, 1, len(runs)))
run_to_color = dict(zip(runs, colors))

# `timestamp` is when the message ARRIVED and stays the join key; `stamp` is when the pose
# was true, and is the only one an interval may be computed from.
TIME = pose_time_base(df)
if TIME != 'stamp':
    print("No publisher timestamps in this campaign: intervals below are arrival times, "
          "quantized by /clock and jittered by delivery.")

NAV_BEHAVIOR = navigation_behavior(df_behaviors)
print(f"{len(runs)} run(s); navigation behaviour {NAV_BEHAVIOR}.")
df_behavior_info = get_behavior_info(NAV_BEHAVIOR, df_behaviors)
_timed = df_behavior_info['run_id'].nunique() if not df_behavior_info.empty else 0
if _timed < len(runs):
    # An instance is only counted once it reached a terminal status, so a run that was
    # killed mid-navigation contributes none. Named here because every duration panel
    # below is then drawn over fewer runs than the campaign has.
    print(f"{_timed} of {len(runs)} run(s) completed a navigation: the duration "
          f"panels cover those.")
```

```python
# Prepare data for multi-run comparison
import numpy as np
import matplotlib.pyplot as plt

# Split by frame and prepare for all runs
df_robot = df[df['frame'] == 'base_link'].copy()
df_groundtruth = df[df['frame'] == 'nav2_turtlebot4_base_link_gt'].copy()

# Sort by run and timestamp
# Sorted on TIME: ordering by arrival time leaves rows within one tick in arbitrary
# order, which is enough to hand interp1d a non-monotonic x.
df_robot.sort_values(['run_id', TIME], inplace=True)
df_groundtruth.sort_values(['run_id', TIME], inplace=True)

# print(f"Processing {len(runs)} runs for comparison")
# print(f"Robot data points: {len(df_robot)}")
# print(f"Ground truth data points: {len(df_groundtruth)}")
```

```python
# Calculate errors for each run separately
from scipy.interpolate import interp1d

# Store results for each run
run_errors = {}
run_stats = []

for run in runs:
    # Get data for this specific run
    df_robot_run = df_robot[df_robot['run_id'] == run].copy()
    df_gt_run = df_groundtruth[df_groundtruth['run_id'] == run].copy()
    if df_robot_run.empty or df_gt_run.empty:
        print(f"Skipping run {run} due to empty data")
        continue
    # Find common time range
    time_start = max(df_robot_run[TIME].iloc[0], df_gt_run[TIME].iloc[0])
    time_end = min(df_robot_run[TIME].iloc[-1], df_gt_run[TIME].iloc[-1])
    
    # Create interpolation functions for ground truth
    gt_timestamps = df_gt_run[TIME].values
    gt_interp_x = interp1d(gt_timestamps, df_gt_run['position.x'].values, kind='linear', fill_value='extrapolate')
    gt_interp_y = interp1d(gt_timestamps, df_gt_run['position.y'].values, kind='linear', fill_value='extrapolate')
    gt_interp_yaw = interp1d(gt_timestamps, df_gt_run['orientation.yaw'].values, kind='linear', fill_value='extrapolate')
    
    # Filter robot data to common time range
    time_mask = (df_robot_run[TIME] >= time_start) & (df_robot_run[TIME] <= time_end)
    df_robot_aligned = df_robot_run[time_mask].copy()
    
    robot_timestamps = df_robot_aligned[TIME].values
    
    # Interpolate ground truth at robot timestamps
    gt_x = gt_interp_x(robot_timestamps)
    gt_y = gt_interp_y(robot_timestamps)
    gt_yaw = gt_interp_yaw(robot_timestamps)
    
    # Calculate position errors
    position_error_x = df_robot_aligned['position.x'].values - gt_x
    position_error_y = df_robot_aligned['position.y'].values - gt_y
    absolute_position_error = np.sqrt(position_error_x**2 + position_error_y**2)
    
    # Calculate orientation error
    orientation_error = df_robot_aligned['orientation.yaw'].values - gt_yaw
    orientation_error = np.arctan2(np.sin(orientation_error), np.cos(orientation_error))
    
    # Store errors for this run
    run_errors[run] = {
        'timestamps': robot_timestamps,
        'position_error': absolute_position_error,
        'position_error_x': position_error_x,
        'position_error_y': position_error_y,
        'orientation_error': orientation_error
    }
    
    # Calculate statistics
    run_stats.append({
        'run_id': run,
        'mean_pos_error': np.mean(absolute_position_error),
        'std_pos_error': np.std(absolute_position_error),
        'max_pos_error': np.max(absolute_position_error),
        'median_pos_error': np.median(absolute_position_error),
        'p95_pos_error': np.percentile(absolute_position_error, 95),
        'mean_orient_error': np.mean(np.abs(orientation_error)),
        'std_orient_error': np.std(orientation_error),
        'max_orient_error': np.max(np.abs(orientation_error))
    })

# Convert to DataFrame for easy analysis
stats_df = pd.DataFrame(run_stats)
```

```python
error_matrix = []
# The runs that produced an error series, in run order. Not `runs`: cell 6 skips a run that
# recorded only one of the two frames, so indexing `run_errors` by every run raised KeyError
# on any configuration where a trial never moved. The guard here used to be `if run in
# run_errors`, reading cell 6's leftover loop variable -- true whenever the LAST run
# happened to succeed, whatever the others did.
error_runs = [r for r in runs if r in run_errors]
if len(error_runs) < len(runs):
    print(f"{len(runs) - len(error_runs)} of {len(runs)} run(s) have no position-error "
          f"series (one of base_link / ground truth was not recorded) and are absent from "
          f"the panels below.")
if error_runs:
    # Multi-Run Comparison: Position Error Over Time
    fig, axes = plt.subplots(2, 2, figsize=(16, 12))

    # 1. Position errors over time for all runs
    ax = axes[0, 0]
    for run, color in ((r, run_to_color[r]) for r in error_runs):
        errors = run_errors[run]
        # Normalize time to start at 0
        time_normalized = errors['timestamps'] - errors['timestamps'][0]
        ax.plot(time_normalized, errors['position_error'], color=color, alpha=0.6, linewidth=1.5, label=f'Run {run}')

    ax.set_xlabel(f'Time (s, {TIME})', fontsize=11)
    ax.set_ylabel('Absolute Position Error (m)', fontsize=11)
    ax.set_title('Position Error Over Time - All Runs', fontsize=12, fontweight='bold')
    ax.legend(loc='upper right', fontsize=8, ncol=2)
    ax.grid(True, alpha=0.3)

    # 2. Mean error trajectory with standard deviation band
    ax = axes[0, 1]
    # Find common time length (use shortest run)
    min_length = min(len(run_errors[run]['position_error']) for run in error_runs)

    # Interpolate all runs to common time grid
    common_time = np.linspace(0, 100, min_length)  # Normalized to 100 seconds

    for run in error_runs:
        errors = run_errors[run]
        time_normalized = np.linspace(0, 100, len(errors['position_error']))
        interp_func = interp1d(time_normalized, errors['position_error'], kind='linear')
        error_matrix.append(interp_func(common_time))

    error_matrix = np.array(error_matrix)
    mean_error = np.mean(error_matrix, axis=0)
    std_error = np.std(error_matrix, axis=0)

    ax.plot(common_time, mean_error, 'b-', linewidth=2.5, label='Mean')
    ax.fill_between(common_time, mean_error - std_error, mean_error + std_error, 
                    alpha=0.3, color='blue', label='±1 Std Dev')
    ax.set_xlabel(f'Normalized Time (s, {TIME})', fontsize=11)
    ax.set_ylabel('Absolute Position Error (m)', fontsize=11)
    ax.set_title('Mean Position Error with Variability Band', fontsize=12, fontweight='bold')
    ax.legend(fontsize=10)
    ax.grid(True, alpha=0.3)

    # 3. Orientation errors over time for all runs
    ax = axes[1, 0]
    for run, color in ((r, run_to_color[r]) for r in error_runs):
        errors = run_errors[run]
        time_normalized = errors['timestamps'] - errors['timestamps'][0]
        ax.plot(time_normalized, np.abs(errors['orientation_error']), color=color, alpha=0.6, linewidth=1.5, label=f'Run {run}')

    ax.set_xlabel(f'Time (s, {TIME})', fontsize=11)
    ax.set_ylabel('Absolute Orientation Error (rad)', fontsize=11)
    ax.set_title('Orientation Error Over Time - All Runs', fontsize=12, fontweight='bold')
    ax.legend(loc='upper right', fontsize=8, ncol=2)
    ax.grid(True, alpha=0.3)

    # 4. Box plot comparison of position errors across runs
    ax = axes[1, 1]
    error_data = [run_errors[run]['position_error'] for run in error_runs]
    bp = ax.boxplot(error_data, tick_labels=[f'{r}' for r in error_runs], patch_artist=True)

    # Color boxes
    for patch, color in zip(bp['boxes'], [run_to_color[r] for r in error_runs]):
        patch.set_facecolor(color)
        patch.set_alpha(0.6)

    ax.set_xlabel('Run Number', fontsize=11)
    ax.set_ylabel('Absolute Position Error (m)', fontsize=11)
    ax.set_title('Position Error Distribution Across Runs', fontsize=12, fontweight='bold')
    ax.grid(True, alpha=0.3, axis='y')

    plt.tight_layout()
    plt.show()
```

```python
# Multi-Run Speed Comparison - Crisp Visualizations
fig, axes = plt.subplots(2, 3, figsize=(18, 12))

# 1. Linear Speed Distribution Comparison
ax = axes[0, 0]
for run, color in zip(runs, colors):
    df_run_speeds = df_gt_speeds[df_gt_speeds['run_id'] == run]
    ax.hist(df_run_speeds['linear_speed'].values, bins=40, alpha=0.5, 
            color=color, label=f'Run {run}', edgecolor='black', linewidth=0.5)

ax.set_xlabel('Linear Speed (m/s)', fontsize=11, fontweight='bold')
ax.set_ylabel('Frequency', fontsize=11, fontweight='bold')
ax.set_title('Linear Speed Distribution - All Runs', fontsize=12, fontweight='bold')
ax.legend(fontsize=8, ncol=2)
ax.grid(True, alpha=0.3, axis='y', linewidth=0.5)

# 2. Angular Speed Distribution Comparison
ax = axes[0, 1]
for run, color in zip(runs, colors):
    df_run_speeds = df_gt_speeds[df_gt_speeds['run_id'] == run]
    ax.hist(df_run_speeds['angular_speed'].values, bins=40, alpha=0.5, 
            color=color, label=f'Run {run}', edgecolor='black', linewidth=0.5)

ax.axvline(0, color='black', linestyle='-', linewidth=0.5)
ax.set_xlabel('Angular Speed (rad/s)', fontsize=11, fontweight='bold')
ax.set_ylabel('Frequency', fontsize=11, fontweight='bold')
ax.set_title('Angular Speed Distribution - All Runs', fontsize=12, fontweight='bold')
ax.legend(fontsize=8, ncol=2)
ax.grid(True, alpha=0.3, axis='y', linewidth=0.5)

# 3. Box Plot - Linear Speed Comparison
ax = axes[0, 2]
linear_speed_data = [df_gt_speeds[df_gt_speeds['run_id'] == run]['linear_speed'].values 
                     for run in runs]
bp = ax.boxplot(linear_speed_data, tick_labels=[f'{r}' for r in runs], 
                patch_artist=True, widths=0.6)

for patch, color in zip(bp['boxes'], colors):
    patch.set_facecolor(color)
    patch.set_alpha(0.7)
    patch.set_edgecolor('black')
    patch.set_linewidth(1.5)

for element in ['whiskers', 'fliers', 'means', 'medians', 'caps']:
    plt.setp(bp[element], color='black', linewidth=1.5)

ax.set_xlabel('Run Number', fontsize=11, fontweight='bold')
ax.set_ylabel('Linear Speed (m/s)', fontsize=11, fontweight='bold')
ax.set_title('Linear Speed Distribution by Run', fontsize=12, fontweight='bold')
ax.grid(True, alpha=0.3, axis='y', linewidth=0.5)

# 4. Bar Chart - Mean Linear Speed with Error Bars
ax = axes[1, 0]
# The bars below come from `speed_stats_df`, whose rows are the runs that produced that
# statistic -- not necessarily every run of the configuration. Taking the x axis and
# the colours from `runs` instead put 5 positions against 4 heights, which matplotlib
# refuses; and where the counts happened to match it would have mislabelled every bar.
speed_runs = speed_stats_df['run_id'].tolist()
speed_colors = [run_to_color[r] for r in speed_runs]
x_pos = np.arange(len(speed_runs))
if 'mean_linear' in speed_stats_df and 'std_linear' in speed_stats_df:
    linear_means = speed_stats_df['mean_linear'].values
    linear_stds = speed_stats_df['std_linear'].values

    bars = ax.bar(x_pos, linear_means, yerr=linear_stds, capsize=5, 
                alpha=0.75, color=speed_colors, edgecolor='black', linewidth=1.5)
    ax.set_xlabel('Run Number', fontsize=11, fontweight='bold')
    ax.set_ylabel('Mean Linear Speed (m/s)', fontsize=11, fontweight='bold')
    ax.set_title('Mean Linear Speed by Run', fontsize=12, fontweight='bold')
    ax.set_xticks(x_pos)
    ax.set_xticklabels([f'{r}' for r in speed_runs])
    ax.grid(True, alpha=0.3, axis='y', linewidth=0.5)

    # Add value labels
    for i, (mean, std) in enumerate(zip(linear_means, linear_stds)):
        ax.text(i, mean + std + 0.01, f'{mean:.3f}', ha='center', va='bottom', 
                fontsize=8, fontweight='bold')

# 5. Bar Chart - Mean Angular Speed with Error Bars
ax = axes[1, 1]
if 'mean_angular' in speed_stats_df and 'std_angular' in speed_stats_df:
    angular_means = speed_stats_df['mean_angular'].values
    angular_stds = speed_stats_df['std_angular'].values

    bars = ax.bar(x_pos, angular_means, yerr=angular_stds, capsize=5, 
                alpha=0.75, color=speed_colors, edgecolor='black', linewidth=1.5)
    ax.set_xlabel('Run Number', fontsize=11, fontweight='bold')
    ax.set_ylabel('Mean Angular Speed (rad/s)', fontsize=11, fontweight='bold')
    ax.set_title('Mean Angular Speed by Run', fontsize=12, fontweight='bold')
    ax.set_xticks(x_pos)
    ax.set_xticklabels([f'{r}' for r in speed_runs])
    ax.grid(True, alpha=0.3, axis='y', linewidth=0.5)

    # Add value labels
    for i, (mean, std) in enumerate(zip(angular_means, angular_stds)):
        ax.text(i, mean + std + 0.01, f'{mean:.3f}', ha='center', va='bottom', 
                fontsize=8, fontweight='bold')

# 6. Scatter Plot - Max Speed Comparison
ax = axes[1, 2]
if 'max_linear' in speed_stats_df and 'max_angular' in speed_stats_df:
    max_linear = speed_stats_df['max_linear'].values
    max_angular = speed_stats_df['max_angular'].values

    scatter = ax.scatter(max_linear, max_angular, c=speed_colors, s=250, alpha=0.75, 
                        edgecolors='black', linewidth=2)

    for i, run in enumerate(runs):
        ax.annotate(f'Run {run}', (max_linear[i], max_angular[i]),
                xytext=(5, 5), textcoords='offset points', fontsize=9, fontweight='bold')

    ax.set_xlabel('Max Linear Speed (m/s)', fontsize=11, fontweight='bold')
    ax.set_ylabel('Max Angular Speed (rad/s)', fontsize=11, fontweight='bold')
    ax.set_title('Maximum Speed Comparison', fontsize=12, fontweight='bold')
    ax.grid(True, alpha=0.3, linewidth=0.5)

plt.tight_layout()
plt.show()
```

```python
# Speed Profile Over Time - Multi-Run Comparison
fig, axes = plt.subplots(2, 1, figsize=(16, 10))

# 1. Linear Speed Over Time - All Runs
ax = axes[0]
for run, color in zip(runs, colors):
    df_run_speeds = df_gt_speeds[df_gt_speeds['run_id'] == run]
    timestamps = df_run_speeds[TIME].values
    if len(timestamps) == 0:
        print(f"Warning: No speed data for run {run}. Skipping.")
        continue
    # Normalize time to start at 0
    timestamps_norm = timestamps - timestamps[0]
    linear_speeds = df_run_speeds['linear_speed'].values
    
    ax.plot(timestamps_norm, linear_speeds, color=color, alpha=0.7, 
            linewidth=1.5, label=f'Run {run}')

ax.set_xlabel(f'Time (s, {TIME})', fontsize=12, fontweight='bold')
ax.set_ylabel('Linear Speed (m/s)', fontsize=12, fontweight='bold')
ax.set_title('Linear Speed Profile Over Time - All Runs', fontsize=13, fontweight='bold')
ax.legend(loc='upper right', fontsize=9, ncol=3)
ax.grid(True, alpha=0.3, linewidth=0.5)
ax.set_ylim(bottom=0)

# 2. Angular Speed Over Time - All Runs
ax = axes[1]
for run, color in zip(runs, colors):
    df_run_speeds = df_gt_speeds[df_gt_speeds['run_id'] == run]
    timestamps = df_run_speeds[TIME].values
    if len(timestamps) == 0:
        print(f"Warning: No speed data for run {run}. Skipping.")
        continue
    timestamps_norm = timestamps - timestamps[0]
    angular_speeds = df_run_speeds['angular_speed'].values
    
    ax.plot(timestamps_norm, angular_speeds, color=color, alpha=0.7, 
            linewidth=1.5, label=f'Run {run}')

ax.axhline(0, color='black', linestyle='-', linewidth=0.5)
ax.set_xlabel(f'Time (s, {TIME})', fontsize=12, fontweight='bold')
ax.set_ylabel('Angular Speed (rad/s)', fontsize=12, fontweight='bold')
ax.set_title('Angular Speed Profile Over Time - All Runs', fontsize=13, fontweight='bold')
ax.legend(loc='upper right', fontsize=9, ncol=3)
ax.grid(True, alpha=0.3, linewidth=0.5)

plt.tight_layout()
plt.show()
```
